```python
# Експеримент: каталог навчальних відео
# Мета: перетворення тексту на структуровані записи.
# Поля: title, topic, duration_minutes.
# Усі поля обов'язкові.
# Тривалість має бути додатним цілим числом.
```

```python
cases = [
    {
        "text": "Відео Python для початківців, тема цикли, тривалість 15 хвилин",
        "expected": {
            "title": "Python для початківців",
            "topic": "цикли",
            "duration_minutes": 15
        }
    },
    {
        "text": "Урок HTML, тема форми, тривалість 20 хвилин",
        "expected": {
            "title": "HTML",
            "topic": "форми",
            "duration_minutes": 20
        }
    },
    {
        "text": "Відео про CSS, тема Flexbox",
        "expected": None
    },
]

print("Кількість тестових випадків:", len(cases))
for i, case in enumerate(cases, start=1):
    print(f"{i}. {case['text']}")
```

```python
cases.extend([
    {
        "text": "Відео JavaScript, тема функції, тривалість 25 хвилин",
        "expected": {
            "title": "JavaScript",
            "topic": "функції",
            "duration_minutes": 25
        }
    },
    {
        "text": "Урок SQL, тема бази даних, тривалість 30 хвилин",
        "expected": {
            "title": "SQL",
            "topic": "бази даних",
            "duration_minutes": 30
        }
    },
    {
        "text": "Відео Git, тема коміти, тривалість 10 хвилин",
        "expected": {
            "title": "Git",
            "topic": "коміти",
            "duration_minutes": 10
        }
    },
    {
        "text": "Відео Python, тема списки",
        "expected": None
    },
    {
        "text": "Урок HTML, тривалість 15 хвилин",
        "expected": None
    },
    {
        "text": "Відео CSS, тема Grid, тривалість -5 хвилин",
        "expected": None
    },
    {
        "text": "Відео Java, тема класи, тривалість 0 хвилин",
        "expected": None
    }
])

print("Загальна кількість тестових випадків:", len(cases))
```

```python
def validate_video(video):
    if not isinstance(video, dict):
        return False

    title = video.get("title")
    topic = video.get("topic")
    duration = video.get("duration_minutes")

    if not isinstance(title, str) or not title.strip():
        return False

    if not isinstance(topic, str) or not topic.strip():
        return False

    if type(duration) is not int or duration <= 0:
        return False

    return True


for i, case in enumerate(cases, start=1):
    expected = case["expected"]

    if expected is None:
        print(f"{i}. Потрібне уточнення")
    elif validate_video(expected):
        print(f"{i}. Дані правильні")
    else:
        print(f"{i}. Помилка в даних")
```

```python
import re

def rule_based_extract(text):
    pattern = (
        r"(?:Відео|Урок)\s+(.+?),\s*"
        r"тема\s+(.+?),\s*"
        r"тривалість\s+(-?\d+)\s+хвилин"
    )

    match = re.fullmatch(pattern, text.strip(), re.IGNORECASE)

    if not match:
        return None

    video = {
        "title": match.group(1).strip(),
        "topic": match.group(2).strip(),
        "duration_minutes": int(match.group(3))
    }

    if validate_video(video):
        return video

    return None


for i, case in enumerate(cases, start=1):
    result = rule_based_extract(case["text"])
    print(f"{i}. {result}")
```

```python
correct = 0

for i, case in enumerate(cases, start=1):
    result = rule_based_extract(case["text"])
    expected = case["expected"]

    if result == expected:
        print(f"{i}. PASS")
        correct += 1
    else:
        print(f"{i}. FAIL")

print()
print(f"Правильних відповідей: {correct}/{len(cases)}")
print(f"Точність: {correct / len(cases) * 100:.1f}%")

assert correct == len(cases), "Не всі тести пройдено"
```

```python
import json

def llm_extract(text):
    prompt = f"""
Витягни інформацію про навчальне відео з тексту.

Потрібні поля:
- title: назва відео
- topic: тема
- duration_minutes: тривалість у хвилинах (ціле додатне число)

Правила:
- Не вигадуй жодних даних.
- Назва йде після слова "Відео" або "Урок".
- Слова "Відео" та "Урок" не включай у назву.
- Тему бери тільки після слова "тема".
- Тривалість бери тільки після слова "тривалість".
- Якщо назва, тема або тривалість відсутні, поверни null.
- Якщо тривалість менша або дорівнює нулю, поверни null.
- Повертай тільки JSON або null, без пояснень.

Текст: {json.dumps(text, ensure_ascii=False)}
"""

    try:
        response = client.chat.completions.create(
            model="openai/gpt-oss-20b",
            messages=[
                {
                    "role": "system",
                    "content": "Ти витягуєш структуровані дані. Відповідай лише JSON."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=0
        )

        result = json.loads(response.choices[0].message.content)

        if validate_video(result):
            return result

        return None

    except (json.JSONDecodeError, ValueError, TypeError):
        return None
```

```python
text = "Відео Python для початківців, тема цикли, тривалість 15 хвилин"

result = llm_extract(text)

print("Результат ШІ:", result)
```

```python
rule_correct = 0
llm_correct = 0

for i, case in enumerate(cases, start=1):
    text = case["text"]
    expected = case["expected"]

    rule_result = rule_based_extract(text)
    llm_result = llm_extract(text)

    rule_ok = rule_result == expected
    llm_ok = llm_result == expected

    if rule_ok:
        rule_correct += 1

    if llm_ok:
        llm_correct += 1

    print(f"Тест {i}:")
    print(f"  Очікувано: {expected}")
    print(f"  Rule-based: {rule_result} — {'PASS' if rule_ok else 'FAIL'}")
    print(f"  LLM:        {llm_result} — {'PASS' if llm_ok else 'FAIL'}")
    print()

print("ПІДСУМКИ:")
print(f"Rule-based: {rule_correct}/{len(cases)}")
print(f"LLM:        {llm_correct}/{len(cases)}")
```
