```python
import requests

BASE_URL = "http://127.0.0.1:8000"

response = requests.get(f"{BASE_URL}/videos")

print("Статус:", response.status_code)
print("Відео:", response.json())
```

```python
video_data = {
    "title": "Python для початківців",
    "topic": "Цикли",
    "duration_minutes": 15
}

response = requests.post(
    f"{BASE_URL}/videos",
    json=video_data
)

print("Статус:", response.status_code)
print("Створене відео:", response.json())
```

```python
invalid_video = {
    "title": "Python",
    "topic": "Цикли",
    "duration_minutes": -5
}

response = requests.post(
    f"{BASE_URL}/videos",
    json=invalid_video
)

print("Статус:", response.status_code)
print("Помилка:", response.json())
```

```python
ai_data = {
    "text": "Відео Python для початківців, тема цикли, тривалість 15 хвилин"
}

response = requests.post(
    f"{BASE_URL}/videos/ai",
    json=ai_data
)

print("Статус:", response.status_code)
print("Результат:", response.json())
```

```python
incomplete_data = {
    "text": "Відео Python, тема цикли"
}

response = requests.post(
    f"{BASE_URL}/videos/ai",
    json=incomplete_data
)

print("Статус:", response.status_code)
print("Результат:", response.json())
```

```python
response = requests.get(f"{BASE_URL}/videos")

print("Статус:", response.status_code)
print("Список відео:")

for video in response.json():
    print(video)
```
